### Data Cleaning

```python
# Create a copy before cleaning

df = merged_data.copy()

print("Working dataset created!")
print("Shape:", df.shape)
```

```python
# Check duplicate rows

print("Duplicate rows:", df.duplicated().sum())
```

```python
# Check missing values

missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
# Check important numeric columns

print(df[[
    "budget",
    "revenue",
    "popularity",
    "runtime",
    "vote_average",
    "vote_count"
]].describe())
```

```python
# Convert release_date to datetime

df["release_date"] = pd.to_datetime(
    df["release_date"],
    errors="coerce"
)

print(df["release_date"].dtype)
```

```python
# Extract release year

df["release_year"] = df["release_date"].dt.year

print(df[["title", "release_date", "release_year"]].head())
```

# Handle missing numeric values

```python
# Fill missing numerical values with median

numeric_columns = [
    "budget",
    "revenue",
    "popularity",
    "runtime",
    "vote_average",
    "vote_count"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing numerical values handled.")
```

```python
# Check missing values in important text columns

text_columns = [
    "title",
    "overview",
    "genres",
    "keywords",
    "cast",
    "crew"
]

print(df[text_columns].isnull().sum())
```

```python
# Fill missing text values

for column in text_columns:
    df[column] = df[column].fillna("")

print("Missing text values handled.")
```

```python
# Final missing-value check

print("Remaining missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Check cleaned dataset

print("Final shape:", df.shape)
df.head()
```

```python
# Check whether movie IDs match

id_check = df["id"] == df["movie_id"]

print("Matching IDs:", id_check.sum())
print("Non-matching IDs:", (~id_check).sum())
```

# EXPLORATORY DATA ANALYSIS (EDA)

```python

import ast
import pandas as pd
import numpy as np

# ------------------------------------------
# 1. Basic Dataset Information
# ------------------------------------------

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()


# ------------------------------------------
# 2. Statistical Summary
# ------------------------------------------

print("\nStatistical Summary:")
display(df.describe())


# ------------------------------------------
# 3. Total Number of Movies
# ------------------------------------------

total_movies = df["title"].nunique()

print("\nTotal Number of Unique Movies:", total_movies)


# ------------------------------------------
# 4. Average Movie Rating
# ------------------------------------------

average_rating = df["vote_average"].mean()

print("Average Movie Rating:", round(average_rating, 2))


# ------------------------------------------
# 5. Top 10 Most Popular Movies
# ------------------------------------------

top_popular = (
    df[
        ["title", "popularity", "vote_average", "vote_count"]
    ]
    .sort_values(
        by="popularity",
        ascending=False
    )
    .head(10)
)

print("\nTop 10 Most Popular Movies:")
display(top_popular)


# ------------------------------------------
# 6. Top 10 Highest Rated Movies
# ------------------------------------------

top_rated = (
    df[df["vote_count"] >= 100]
    [
        ["title", "vote_average", "vote_count"]
    ]
    .sort_values(
        by="vote_average",
        ascending=False
    )
    .head(10)
)

print("\nTop 10 Highest Rated Movies:")
display(top_rated)


# ------------------------------------------
# 7. Top 10 Movies by Revenue
# ------------------------------------------

top_revenue = (
    df[
        ["title", "revenue", "vote_average", "popularity"]
    ]
    .sort_values(
        by="revenue",
        ascending=False
    )
    .head(10)
)

print("\nTop 10 Movies by Revenue:")
display(top_revenue)


# ------------------------------------------
# 8. Movies Released Per Year
# ------------------------------------------

movies_per_year = (
    df.dropna(subset=["release_year"])
    .groupby("release_year")
    .size()
    .reset_index(name="movie_count")
)

print("\nMovies Released Per Year:")
display(movies_per_year.head(10))


# ------------------------------------------
# 9. Years With Most Movie Releases
# ------------------------------------------

top_release_years = (
    movies_per_year
    .sort_values(
        by="movie_count",
        ascending=False
    )
    .head(10)
)

print("\nTop 10 Years With Most Movie Releases:")
display(top_release_years)


# ------------------------------------------
# 10. Extract Genres
# ------------------------------------------

def extract_names(value):
    try:
        data = ast.literal_eval(value)

        if isinstance(data, list):
            return [
                item["name"]
                for item in data
                if isinstance(item, dict) and "name" in item
            ]

        return []

    except:
        return []


df["genre_list"] = df["genres"].apply(extract_names)

print("\nSample Genre Data:")
display(df[["title", "genre_list"]].head())


# ------------------------------------------
# 11. Genre Counts
# ------------------------------------------

genre_counts = (
    df["genre_list"]
    .explode()
    .dropna()
    .value_counts()
)

print("\nTop 10 Most Common Genres:")
display(genre_counts.head(10))


# ------------------------------------------
# 12. Average Rating by Genre
# ------------------------------------------

genre_rating = (
    df[["genre_list", "vote_average"]]
    .explode("genre_list")
    .dropna(subset=["genre_list"])
    .groupby("genre_list")["vote_average"]
    .mean()
    .sort_values(ascending=False)
)

print("\nAverage Rating by Genre:")
display(genre_rating.head(10))


# ------------------------------------------
# 13. Correlation Analysis
# ------------------------------------------

numeric_columns = [
    "budget",
    "revenue",
    "popularity",
    "runtime",
    "vote_average",
    "vote_count"
]

correlation = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
display(correlation)


# ------------------------------------------
# 14. Basic Findings
# ------------------------------------------

print("\n========== EDA SUMMARY ==========")

print("Total Movies:", total_movies)

print("Average Rating:", round(average_rating, 2))

print(
    "Most Common Genre:",
    genre_counts.index[0]
    if len(genre_counts) > 0
    else "Not available"
)

print(
    "Most Popular Movie:",
    top_popular.iloc[0]["title"]
    if len(top_popular) > 0
    else "Not available"
)

print(
    "Highest Revenue Movie:",
    top_revenue.iloc[0]["title"]
    if len(top_revenue) > 0
    else "Not available"
)

print("=================================")
```
